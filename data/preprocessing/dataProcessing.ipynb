```python
import pandas as pd
import re
```

```python
# Load the CSV file
df = pd.read_csv('/home/satyamkumarjha/AIML/SIH_2025/AvsarKendra_AI/data/raw/JobDataset.csv')
```

```python
# Task 1: Convert Skills Requirements to list of strings
df['Skills Requirements'] = df['Skills Requirements'].str.split(',')

# Task 2: Extract only numbers from Duration column
df['Duration'] = df['Duration'].str.extract('(\d+)').astype(int)
```

```python
# decalaring the type mapping dictionary
type_mapping = {
    'SDE': 'Software Development',
    'SDE (Security)': 'Software Development',
    'SDE (Mobile)': 'Software Development',
    'SDE (Gaming)': 'Game Development',
    'SDE (QA)': 'Software Testing',
    'SDE (DevOps)': 'DevOps and Cloud',
    'SDE (Blockchain)': 'Blockchain Development',
    'Data Science': 'Data & Analytics',
    'Marketing': 'Business Administration',
    'Core Engineering': 'Engineering',
    'Finance': 'Business Administration',
    'Human Resources': 'Business Administration',
    'Media': 'Creative & Media',
    'Legal': 'Legal & Compliance',
    'Design': 'Creative & Media',
    'Product Management': 'Business Administration',
    'Sales': 'Business Administration',
    'Healthcare': 'Healthcare & Life Sciences',
    'Operations': 'Business Administration',
    'Management': 'Business Administration',
    'Hospitality': 'Business Administration',
}
```

```python
df['Type Category'] = df['Type'].map(type_mapping)
df = df.drop(columns=['Type'])
df = df.rename(columns={'Type Category': 'Type'})
```
