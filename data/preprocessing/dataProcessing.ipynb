```python
import pandas as pd
import re
import uuid
```

```python
# Load the CSV file
df = pd.read_csv('../raw/raw.csv')
```

```python
df.head()
```

```python
type_counts = df['Type'].value_counts(dropna=False)
location_counts = df['Location'].value_counts(dropna=False)
mode_counts = df['Mode'].value_counts(dropna=False)

print("Type bifurcation:\n", type_counts, "\n")
print("Location bifurcation:\n", location_counts, "\n")
print("Mode bifurcation:\n", mode_counts)
```

```python
# Task 1: Convert Skills Requirements to list of strings
df['Skills Requirements'] = df['Skills Requirements'].str.split(',')

# Task 2: Extract only numbers from Duration column
df['Duration'] = df['Duration'].str.extract('(\d+)').astype(int)
```

```python
# decalaring the type mapping dictionary
type_mapping = {
    "Data Science": "SDE",
    "SDE": "SDE",
    "SDE (DevOps)": "CLOUD",
    "Core Engineering": "DATA_ENGINEER",
    "SDE (Security)": "CYBERSECURITY",
    "Product Management": "PRODUCT_MANAGER",
    "IT Operations": "CLOUD",
    "SDE (Mobile)": "SDE",
    "Tech Consulting": "SDE",
    "SDE (Embedded)": "SDE",
    "SDE (Telecom)": "DATA_ENGINEER",
    "SDE (IoT)": "SDE",
    "SDE (Gaming)": "GAME_DEV",
    "Design": "UI_UX",
    "SDE (AI)": "SDE",
    "SDE (Cloud)": "CLOUD",
    "SDE (Research)": "SDE",
    "SDE (AR/VR)": "GAME_DEV",
    "SDE (Firmware)": "SDE",
    "SDE (RPA)": "SDE",
    "SDE (GIS)": "DATA_ENGINEER",
}


```

```python
df['Type Category'] = df['Type'].map(type_mapping)
df = df.drop(columns=['Type'])
df = df.rename(columns={'Type Category': 'JobType'})
```

```python
# Generate random unique IDs
df.insert(0, "JobID", [uuid.uuid4().hex[:8] for _ in range(len(df))])
```

```python
# Display the first few rows to verify changes
print("First 5 rows after processing:")

df.head()
```

```python
df.to_csv('../raw/jobPostings.csv', index=False)
```
